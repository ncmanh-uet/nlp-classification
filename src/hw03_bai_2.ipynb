## Bài 2. Trích xuất tất cả các địa chỉ email
Mẫu tìm phần tên hộp thư trước `@`, rồi đến tên miền và đuôi miền. Cụm `@2PM` không có đủ các phần này nên không khớp.

```python
s = 'Hello from shubhamg199630@gmail.com to priya@yahoo.com      about the meeting @2PM'

email_pattern = r'[a-zA-Z0-9._%+-]+@[a-zA-Z0-9-]+(?:\.[a-zA-Z0-9-]+)*\.[a-zA-Z]{2,}'
emails = re.findall(email_pattern, s)
print(emails)
```
