## Bài 1. Trích xuất tất cả các số
Mẫu `\d+` khớp với một hoặc nhiều chữ số liên tiếp. Giữ kết quả ở dạng chuỗi để bảo toàn số `0` ở đầu số điện thoại.

```python
s = 'My 2 favourite numbers are 8 and 25. My mobile is 0912203062'

numbers = re.findall(r'\d+', s)
print(numbers)
```
